```python
na_dropped = raw_accident_data.dropna()
na_dropped
```

```python
# data cleaning for the NaN data
clean_data = na_dropped.reset_index(drop=True) 
# resetting so the index are starting from 0
accidentsCleanedData = clean_data
clean_data
# displaying all the columns
print(clean_data.columns)
```

```python
# getting the count of severity levels of accidents 
serious_count = 0
fatal_count = 0
slight_count = 0

for severity in clean_data["Accident_Severity"]:
    if severity == "Serious":
        serious_count += 1
    if severity == "Fatal":
        fatal_count += 1
    if severity == "Slight":
        slight_count += 1
        
print("Serious Accident: ",serious_count)
print("Fatal Accident: ",fatal_count)
print("Slight Accident: ",slight_count)
```

```python
# exploring conditions of the road and weather using the cleaned 
dry_count = 0
wet_count = 0

for road_conditions in clean_data["Road_Surface_Conditions"]:
    if road_conditions == "Dry":
        dry_count += 1
    if road_conditions == "Wet or damp":
        wet_count += 1

print("Dry Conditions: ", dry_count)
print("Wet Conditions: ", wet_count)

unique_weather = clean_data["Weather_Conditions"].unique()
print("this is all the unique types of weather: ", unique_weather)

fine_count = 0
rain_count = 0
snow_count = 0

for weather_conditions in clean_data["Weather_Conditions"]:
    if weather_conditions == "Fine no high winds":
        fine_count += 1

print(fine_count)

#create a function to calculate daily average accident count
def average_daily_accidents(clean_data):
    return clean_data.groupby("Accident Date").size().mean()

avg_accidents = average_daily_accidents(clean_data)
print("accidents per day: ", avg_accidents)

#create a function to change the time format to HH:MM AM/PM format
# will use this for seeing trends of accidents during certain times
def TimeToHours():
    clean_data['Time'] = pd.to_datetime(clean_data['Time'], errors='coerce')
    # overwriting the time cells to 12 hour format
    clean_data['Time'] = clean_data['Time'].dt.strftime('%-I %p')

# call the function to make the modification
TimeToHours()
# return a specified number of rows from the top
clean_data.head()

```

```python
import numpy as np

# converting csv columns to numpy arrays
severity = clean_data["Accident_Severity"].to_numpy()
light = clean_data["Light_Conditions"].to_numpy()

# getting unique values for rows and columns
unique_severity = np.unique(severity)     
unique_light = np.unique(light)      
print(unique_light, unique_severity)     

# setting up an empty heatmap
heatmap = np.zeros((len(unique_severity), len(unique_light)))

# filling in the heatmap
# loopings through the array of unique severities 
for i in range(len(unique_severity)):
    # inner looping through the unique light conditions
    # checking every severity through each light condition
    for j in range(len(unique_light)):
        # Count accidents for each severity and light condition
        #np.sum counts the array (true or false) for severity and light condition match
        heatmap[i, j] = np.sum((severity == unique_severity[i]) & (light == unique_light[j]))

# after storing heatmap, set up plot
plt.figure(figsize=(12,5))
# displays 2D data as image, cell color to red, 
plt.imshow(heatmap, cmap="Reds")
# add colorbar for the heat scale bar
plt.colorbar(label="Number of Accidents")
# positioning each x axis label at the correct index and aligning text to the right
plt.xticks(np.arange(len(unique_light)), unique_light, rotation=45, ha="right")
plt.yticks(np.arange(len(unique_severity)), unique_severity)

#setting x and y labels
plt.xlabel("Light Conditions")
plt.ylabel("Accident Severity")
plt.title("Severity vs Light Conditions")
plt.savefig("severityVsLightConditions.png")
plt.show()
```

```python
def accidentsByHour(clean_data, show=False):

    # the time in the Time column header will be converted to time format instead of a string
    clean_data['Time'] = pd.to_datetime(
        clean_data['Time'],
        format='%I %p',
        errors='coerce'
    ).dt.hour

    #print(clean_data['Time'].dtype)

    # assigning the value to calculate the accidents by every hour
    # group by hour of the accident
    # unstack to create one column per severity
    accidents_per_hour_severity = (
        clean_data.groupby(['Time', 'Accident_Severity']).size().unstack(fill_value=0)
    )
    #print(accidents_per_hour_severity)

    # create chart with matplotlib
    fig = plt.figure(figsize=(10, 5))

    # loop through each severity type 
    for severity in accidents_per_hour_severity.columns:
        plt.scatter(
            # at each severity it checks the index for hour
            accidents_per_hour_severity.index,
            # at each severity how many accidents at each hour
            accidents_per_hour_severity[severity],
            # labeling legend by severity type
            label=severity
        )

    # Add gridlines
    plt.grid(True, which='both', linestyle='--', alpha=0.7)

    plt.xlabel('Hour of Day')
    plt.ylabel('Number of Accidents')
    plt.title('Number of Accidents by Hour of Day')
    plt.xticks(range(24))
    plt.legend(title='Accident Severity')    
    plt.savefig("AccidentsByHourOfDay.png")

    if show:
        plt.show()
        return None
    
    return fig
```

```python
import seaborn as sns

# I created a function to reuse the function for GUI tkinter 
def accident_by_roadtypehours_boxplot(clean_data, show=False):

    # Group by Road_Type and Time to count accidents per hour
    accidents_per_hour_road = clean_data.groupby(['Road_Type', 'Time']).size().reset_index(name='Accident_Count')
    # print("this is the new table:\n", accidents_per_hour_road.head()) 

    # Create the boxplot
    
    plt.close('all')
    fig = plt.figure(figsize=(10, 5))
    
    sns.boxplot(
        # compare distribution of value across road type and accidents
        x='Road_Type',
        y='Accident_Count',
        data=accidents_per_hour_road,
        # palette for differentiating colors
        palette='Set2'
    )

    plt.tight_layout()
    plt.title('Distribution of Accidents per Hour by Road Type')
    plt.xlabel('Road Type')
    plt.ylabel('Number of Accidents per Hour')
    plt.xticks(rotation=45)
    if show:
        plt.show()
        return None
    
    return fig

```

```python
import tkinter as tk
from tkinter import ttk
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg

# create the main Tkinter window
root = tk.Tk()
root.title("UK Accident Dashboard")
root.geometry("1000x600")

frame_plot = tk.Frame(root)
frame_plot.pack(side=tk.TOP, fill=tk.BOTH, expand=True)
canvas = None

# seaborn boxplot
def boxplot():
    global canvas
    fig = accident_by_roadtypehours_boxplot(clean_data)
    if canvas:
        canvas.get_tk_widget().destroy()
    canvas = FigureCanvasTkAgg(fig, master=frame_plot)
    canvas.draw()
    canvas.get_tk_widget().pack(fill=tk.BOTH, expand=True)

# matplotlib scatterplot
def scatterplot():
    global canvas
    fig = accidentsByHour(clean_data)
    if canvas:
        canvas.get_tk_widget().destroy()
    canvas = FigureCanvasTkAgg(fig, master=frame_plot)
    canvas.draw()
    canvas.get_tk_widget().pack(fill=tk.BOTH, expand=True)

# buttons for different plots in the bottom of the frame
tk.Button(root, text="Show Accidents By Road Type by Hour of Day Boxplot", command=boxplot).pack(pady=10)
tk.Button(root, text="Show Accidents By Hour Scatterplot", command=scatterplot).pack(pady=10)

root.mainloop()

```
