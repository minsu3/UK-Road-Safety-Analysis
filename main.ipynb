```python
df_dropped = raw_accident_data.dropna()
# data cleaning for the NaN data
df_dropped
```

```python
df_cleaned = df_dropped.reset_index(drop=True) # resetting so the index are starting from 0
accidentsCleanedData = df_cleaned
```

```python
df_cleaned

# getting the count of severity levels of accidents 
serious_count = 0
fatal_count = 0
slight_count = 0

for severity in df_cleaned["Accident_Severity"]:
    if severity == "Serious":
        serious_count += 1
    if severity == "Fatal":
        fatal_count += 1
    if severity == "Slight":
        slight_count += 1
        
print("Serious Accident: ",serious_count)
print("Fatal Accident: ",fatal_count)
print("Slight Accident: ",slight_count)

```

```python
# exploring conditions of the road and weather

dry_count = 0
wet_count = 0

for road_conditions in df_cleaned["Road_Surface_Conditions"]:
    if road_conditions == "Dry":
        dry_count += 1
    if road_conditions == "Wet or damp":
        wet_count += 1

print("Dry Conditions: ", dry_count)
print("Wet Conditions: ", wet_count)


unique_weather = df_cleaned["Weather_Conditions"].unique()
print("this is all the unique types of weather: ", unique_weather)

fine_count = 0
rain_count = 0
snow_count = 0

for weather_conditions in df_cleaned["Weather_Conditions"]:
    if weather_conditions == "Fine no high winds":
        fine_count += 1

print(fine_count)

def average_daily_accidents(df_cleaned):
    return df_cleaned.groupby("Accident Date").size().mean()

avg_accidents = average_daily_accidents(df_cleaned)
print("accidents per day: ", avg_accidents)
```
